```python
resources = [
   [101, "Laptop", "Electronics", 8],
  [102, "Projector", "Electronics", 0],
  [103, "Whiteboard", "Classroom", 5],
  [104, "Microphone", "Audio", 3],
  [105, "Camera", "Electronics", 2]
   ]
```

```python
def display_resources(resources):
    for resource in resources:
        print(resource)
```

```python
def search_by_category(resources, category):
    for resource in resources:
        if resource[2] == category:
            print(resource)
```

```python
def find_unavailable(resources):
    for resource in resources:
        if resource[3] == 0:
            print(resource)
```

```python
print("All Resources:")
display_resources(resources)

print("\nResources in Electronics Category:")
search_by_category(resources, "Electronics")

print("\nUnavailable Resources:")
find_unavailable(resources)
```

```python
students = [
    [201, "Amina", [78, 85, 90]],
    [202, "Rahim", [65, 70, 68]],
    [203, "Nadia", [92, 88, 95]],
    [204, "Hasan", [55, 60, 65]],
    [205, "Tania", [80, 80, 80]]
]
```

```python
def calculate_average(marks):
    average = sum(marks) / len(marks)
    return average
```

```python
def classify_performance(average):
    if average >= 80:
        return "Excellent"
    elif average >= 70:
        return "Good"
    elif average >= 60:
        return "Satisfactory"
    else:
        return "Needs Improvement"
```

```python
def display_report(students):
    for student in students:
        student_id = student[0]
        name = student[1]
        marks = student[2]

        average = calculate_average(marks)
        performance = classify_performance(average)

        print("ID:", student_id)
        print("Name:", name)
        print("Average:", round(average, 2))
        print("Performance:", performance)
```

```python
display_report(students)
```

```python
marks = [80, 80, 80]

average = calculate_average(marks)

print("Average:", average)
print("Performance:", classify_performance(average))
```

```python
student1 = [80, 80, 80]
student2 = [75, 85, 80]

avg1 = calculate_average(student1)
avg2 = calculate_average(student2)

print("Student 1 Average:", avg1)
print("Student 2 Average:", avg2)

if avg1 == avg2:
    print("Both students have tied averages")
```

```python
marks = [90, 110, 80]

average = calculate_average(marks)

if average is None:
    print("Invalid Marks")
else:
    print("Invalid mark was not detected")
```
