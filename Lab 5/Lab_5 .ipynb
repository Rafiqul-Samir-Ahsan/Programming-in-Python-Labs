```python
def display_resources(resources):
    if len(resources) == 0:
        print("No resources available")
        return

    for resource in resources:
        if len(resource) != 4:
            print("Incomplete resource information")
            continue

        if type(resource[3]) != int or resource[3] < 0:
            print("Invalid quantity for", resource[1])
            continue

        print(resource)
```

```python
def find_unavailable(resources):
    if len(resources) == 0:
        print("No resources available")
        return

    count = 0
    valid = 0

    for resource in resources:
        if len(resource) != 4:
            print("Incomplete resource information")
            continue

        if type(resource[3]) != int or resource[3] < 0:
            print("Invalid quantity")
            continue

        valid += 1

        if resource[3] == 0:
            print(resource[1], "is unavailable")
            count += 1

    if valid == 0:
        print("No valid resources")
    elif count == 0:
        print("All valid resources are available")
```

```python
def search_by_category(resources, category):
    if len(resources) == 0:
        print("No resources available")
        return

    if type(category) != str or category.strip() == "":
        print("Invalid category")
        return

    found = False

    for resource in resources:
        if len(resource) != 4:
            print("Incomplete resource information")
            continue

        if type(resource[2]) != str:
            print("Invalid category")
            continue

        if type(resource[3]) != int or resource[3] < 0:
            print("Invalid quantity")
            continue

        if resource[2].strip().lower() == category.strip().lower():
            print(resource)
            found = True

    if found == False:
        print("Category not found")
```

```python
def calculate_average(marks):
    if marks is None:
        return "No marks"

    if type(marks) != list:
        return "Invalid marks"

    if len(marks) == 0:
        return "No marks"

    for mark in marks:
        if type(mark) not in (int, float) or not (0 <= mark <= 100):
            return "Invalid marks"

    return sum(marks) / len(marks)
```

```python
def classify_performance(average):
    if average == "Invalid marks" or average == "No marks":
        return average

    if type(average) not in (int, float) or not (0 <= average <= 100):
        return "Invalid average"

    if average >= 80:
        return "Excellent"
    elif average >= 70:
        return "Good"
    elif average >= 60:
        return "Satisfactory"
    else:
        return "Needs Improvement"
```

```python
def display_report(students):
    if type(students) != list or len(students) == 0:
        print("No student records available")
        return

    for student in students:
        if type(student) != list or len(student) != 3:
            print("Incomplete student information")
            continue

        student_id = student[0]
        name = student[1]
        marks = student[2]

        average = calculate_average(marks)
        performance = classify_performance(average)

        print("ID:", student_id)
        print("Name:", name)

        if isinstance(average, str):
            print("Average:", average)
        else:
            print("Average:", round(average, 2))

        print("Performance:", performance)
        print()
```

```python
marks1 = [80, 80, 80]

average = calculate_average(marks1)
print(average)
print(classify_performance(average))

students1 = [[206, "Rafi", marks1]]
display_report(students1)
```
