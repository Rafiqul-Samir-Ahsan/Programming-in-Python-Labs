**Sum of prime numbers below 1000**

```python
total = 0
for number in range(2, 1000):
    prime = True
    for i in range(2, int(number ** 0.5) + 1):
        if number % i == 0:
            prime = False
            break
    if prime:
        total += number
print("Sum:", total)
```

**Fibonacci numbers below N**

```python
n = int(input("Enter N: "))
a, b = 0, 1
while a < n:
    print(a)
    a, b = b, a + b
```
